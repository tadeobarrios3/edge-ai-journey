```python
lecturas_ruido = [30, 32, 85, 90, 28]
alertas = 0
for db in lecturas_ruido:
    if db > 70:
        alertas = alertas + 1
print(alertas)
```

```python
temperaturas = [22, 23, 25, 29, 31, 33, 28, 24, 21, 35]
sobrecalentamientos = 0
suma_total = 0
for temps in temperaturas:
    suma_total = suma_total + temps
    promedio = suma_total / len(temperaturas)
    if temps > 30:
        sobrecalentamientos = sobrecalentamientos + 1
print("La temperatura promedio es de", promedio, "grados")
if sobrecalentamientos >= 3:
  print("El sistema necesita mantenimiento")
```
