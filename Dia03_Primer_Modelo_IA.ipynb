```python
from sklearn.tree import DecisionTreeClassifier

# 1. Características (Features): [temperatura, rpm]
datos_entrenamiento = [
    [40, 1500], [45, 1480], [60, 1450], [70, 1400], # Motores sanos
    [72, 900],  [75, 800],  [82, 600],  [90, 200]   # Motores a punto de fallar
]

# 2. Etiquetas (Labels): 0 = Sano, 1 = Falla inminente
etiquetas = [0, 0, 0, 0, 1, 1, 1, 1]

# 3. Creamos y entrenamos el cerebro de IA
modelo = DecisionTreeClassifier()
modelo.fit(datos_entrenamiento, etiquetas)

# 4. Le pasamos un dato NUEVO que nunca ha visto (ej. 74°C y 850 RPM)
prediccion = modelo.predict([[74, 850]])
print("Predicción de la IA (0=Sano, 1=Falla):", prediccion[0])
```

```python
from sklearn.tree import DecisionTreeClassifier

# 1. Características (Features): [temperatura, rpm, vibracion]
datos_entrenamiento = [
    [40, 1500, 1.2], [45, 1500, 1.2], [35, 1300, 1.0], [37, 1250, 1.4], # Motores sanos
    [75, 800, 8.5], [60, 1500, 10], [90, 1000, 3], [45, 1100, 11]  # Motores a punto de fallar
]

# 2. Etiquetas (Labels): 0 = Sano, 1 = Falla inminente
etiquetas = [0, 0, 0, 0, 1, 1, 1, 1]

# 3. Creamos y entrenamos el cerebro de IA
modelo = DecisionTreeClassifier(random_state=42)
modelo.fit(datos_entrenamiento, etiquetas)

# 4. Le pasamos un dato NUEVO que nunca ha visto (ej. 55°C, 1420 RPM, 9.1 m/s)
prediccion = modelo.predict([[55, 1420, 9.1]])
print("Predicción de la IA (0=Sano, 1=Falla):", prediccion[0])
```
