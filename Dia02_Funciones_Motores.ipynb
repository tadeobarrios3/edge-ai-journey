```python
def diagnosticar_motor(nombre, temperaturas, limite_temp, rpm):
    sobrecalentamientos = 0
    for t in temperaturas:
      if t > limite_temp and rpm == 0:
        return f"EMERGENCIA CRÍTICA en {nombre}: Motor atascado"
      if t > limite_temp:
            sobrecalentamientos += 1

    if sobrecalentamientos >= 3:
        return f"ALERTA en {nombre}: Requiere mantenimiento"
    else:
        return f"OK en {nombre}: Funcionando a la perfección"
```

```python
def diagnosticar_motor(nombre, lecturas, limite_temp):
    sobrecalentamientos = 0
    for dato in lecturas:
      if dato["temp"] > limite_temp and dato["rpm"] == 0:
        return f"EMERGENCIA CRÍTICA en {nombre}: Motor atascado"
      if dato["temp"] > limite_temp:
            sobrecalentamientos += 1

    if sobrecalentamientos >= 3:
        return f"ALERTA en {nombre}: Requiere mantenimiento"
    else:
        return f"OK en {nombre}: Funcionando a la perfección"
print(diagnosticar_motor("Bomba 1", [
    {"temp": 40, "rpm": 1500},
    {"temp": 42, "rpm": 1480},
    {"temp": 44, "rpm": 1510},
    {"temp": 41, "rpm": 1490}
], 45))
print(diagnosticar_motor("Motor Principal", [
    {"temp": 80, "rpm": 1200},
    {"temp": 86, "rpm": 1150},
    {"temp": 88, "rpm": 1100},
    {"temp": 83, "rpm": 1180},
    {"temp": 90, "rpm": 1050}
], 85))
print(diagnosticar_motor("Compresor", [
    {"temp": 60, "rpm": 1500},
    {"temp": 75, "rpm": 800},
    {"temp": 95, "rpm": 0}
], 85))
```
